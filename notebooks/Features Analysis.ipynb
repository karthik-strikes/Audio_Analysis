# Audio Features Analysis - Klaatch Dataset

This notebook performs comprehensive linguistic and audio feature analysis using the DLATK (Distributed Language Analysis ToolKit) framework.

## Objectives
- Extract n-gram features (1-gram, 2-gram, 3-gram) from audio transcriptions
- Perform topic modeling and LDA analysis
- Analyze features across different demographic groups (gender, race)
- Build predictive models for various outcomes

## Dataset
- **Source**: Klaatch audio recordings with transcriptions
- **Features**: Text transcriptions, demographic information (gender, race, age, education), CEL scores
